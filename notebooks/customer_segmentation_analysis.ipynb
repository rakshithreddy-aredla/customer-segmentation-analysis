# Customer Segmentation Analysis\n\nUnsupervised segmentation using K-means clustering to identify distinct customer groups based on purchasing behavior.

```python
import pandas as pd\nimport numpy as np\nimport matplotlib.pyplot as plt\nimport seaborn as sns\nfrom sklearn.cluster import KMeans\nfrom sklearn.preprocessing import StandardScaler\n%matplotlib inline\nsns.set_style('whitegrid')
```

```python
df = pd.read_csv('../data/customers.csv')\ndf.head()
```

```python
df.describe()
```

## Feature Selection & Scaling

```python
features = df[['annual_income_k','spending_score','num_purchases','avg_purchase_value']].copy()\nscaler = StandardScaler()\nX_scaled = scaler.fit_transform(features)
```

## Elbow Method to Find Optimal Clusters

```python
inertia = []\nfor k in range(1,11):\n    km = KMeans(n_clusters=k, n_init='auto', random_state=42)\n    km.fit(X_scaled)\n    inertia.append(km.inertia_)\nplt.figure(figsize=(8,4))\nplt.plot(range(1,11), inertia, marker='o')\nplt.xlabel('Number of clusters (k)')\nplt.ylabel('Inertia')\nplt.title('Elbow Method')\nplt.show()
```

## K-means Clustering (k=4)

```python
kmeans = KMeans(n_clusters=4, n_init='auto', random_state=42)\nclusters = kmeans.fit_predict(X_scaled)\ndf['cluster'] = clusters
```

```python
plt.figure(figsize=(9,6))\nsns.scatterplot(data=df, x='annual_income_k', y='spending_score', hue='cluster', palette='Set2', s=80)\nplt.title('Customer Segments (Income vs Spending Score)')\nplt.show()
```

```python
df.groupby('cluster')[['annual_income_k','spending_score','num_purchases','avg_purchase_value','loyalty_years']].mean().round(2)
```
